```python
import timeit
import numpy as np
from vec import Vec

sizes = [2000, 4000, 8000, 16000, 32000, 64000]

for n in sizes:
    v1 = Vec.uniform(n)
    v2 = Vec.uniform(n)

    a = np.array(v1.elements)
    b = np.array(v2.elements)

    vec_time = timeit.timeit(
        lambda: v1 + v2,
        number=100
    ) / 100

    numpy_time = timeit.timeit(
        lambda: a + b,
        number=100
    ) / 100

    print(
        f"{n:6d}  "
        f"Vec: {vec_time:.6f}s  "
        f"NumPy: {numpy_time:.6f}s"
    )
```
